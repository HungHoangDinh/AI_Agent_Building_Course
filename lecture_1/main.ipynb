### Calling LLM (Large Language Model) using the OpenAI library

````python
client = OpenAI(
    api_key=settings.LLM_API_KEY,
    base_url=settings.LLM_BASE_URL
)

response = client.chat.completions.create(
    model=settings.LLM_CHAT_MODEL,
    messages=[
        {   "role": "system",
            "content": "You are a helpful assistant."
        },
        {
            "role": "user",
            "content": "Explain to me how AI works"
        }
    ]
)

print(response.choices[0].message.content)
````

**Streaming**


````python
client = OpenAI(
    api_key=settings.LLM_API_KEY,
    base_url=settings.LLM_BASE_URL
)

response = client.chat.completions.create(
  model=settings.LLM_CHAT_MODEL,
  messages=[
    {
        "role": "system",
        "content": "You are a helpful assistant."
    },
    {   "role": "user",
        "content": "You can review about Rikkeisoft"
    }
  ],
  stream=True
)

for chunk in response:
    print(chunk.choices[0].delta.content,end="", flush=True)
````

**Structure Output**

````python
from pydantic import BaseModel
client = OpenAI(
    api_key=settings.LLM_API_KEY,
    base_url=settings.LLM_BASE_URL
)

class CalendarEvent(BaseModel):
    name: str
    date: str
    participants: list[str]

completion = client.beta.chat.completions.parse(
    model=settings.LLM_CHAT_MODEL,
    messages=[
        {"role": "system", "content": "Extract the event information."},
        {"role": "user", "content": "John and Susan are going to an AI conference on Friday."},
    ],
    response_format=CalendarEvent,
)

print(completion.choices[0].message.parsed)
````

### Calling LLM (Large Language Model) using the Google GenAI library

````python
from google import genai
client = genai.Client(api_key=settings.LLM_API_KEY)
conversation_history = [
    {
        "role": "user",
        "content": "What are the three largest cities in Spain?"
    }
]

response = client.interactions.create(
    model=settings.LLM_CHAT_MODEL,
    input=conversation_history
)
print(response.outputs[-1].text)
````

**Streaming**

````python
from google import genai

client = genai.Client(api_key=settings.LLM_API_KEY)

stream = client.interactions.create(
    model=settings.LLM_CHAT_MODEL,
    input="Giới thiệu về rikkeisoft.",
    stream=True
)

for chunk in stream:
    if chunk.event_type == "content.delta":
        if chunk.delta.type == "text":
            print(chunk.delta.text, end="", flush=True)
        elif chunk.delta.type == "thought":
            print(chunk.delta.thought, end="", flush=True)
    elif chunk.event_type == "interaction.complete":
        print(f"\n\n--- Stream Finished ---")
        print(f"Total Tokens: {chunk.interaction.usage.total_tokens}")
````

**Structure Output**

````python
from google import genai
from pydantic import BaseModel, Field
from typing import Literal, Union
client = genai.Client(api_key=settings.LLM_API_KEY)

class SpamDetails(BaseModel):
    reason: str = Field(description="The reason why the content is considered spam.")
    spam_type: Literal["phishing", "scam", "unsolicited promotion", "other"]

class NotSpamDetails(BaseModel):
    summary: str = Field(description="A brief summary of the content.")
    is_safe: bool = Field(description="Whether the content is safe for all audiences.")

class ModerationResult(BaseModel):
    decision: Union[SpamDetails, NotSpamDetails]

interaction = client.interactions.create(
    model=settings.LLM_CHAT_MODEL,
    input="Moderate the following content: 'Congratulations! You've won a free cruise. Click here to claim your prize: www.definitely-not-a-scam.com'",
    response_format=ModerationResult.model_json_schema(),
)

parsed_output = ModerationResult.model_validate_json(interaction.outputs[-1].text)
print(parsed_output)
````

# Langchain Prompt Templates

### The Problem with Hard-coded Prompt

```python
llm.invoke("Summarize this text in English with a friendly tone...")
```
- Prompt is hard-coded
- Difficult to reuse
- Hard to maintain
- Hard to test or version
### What is prompt template


````python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
template=PromptTemplate.from_template("Introduce about the company {company_name} in {language} language")

prompt=template.format(company_name="Rikkeisoft", language="Vietnamese")

print(prompt)


````

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
def introduce_company(company_name: str, language: str):
    template=PromptTemplate.from_template("Introduce about the company {company_name} in {language} language")
    prompt=template.format(company_name=company_name, language=language)
    llm=ChatOpenAI(model=settings.LLM_CHAT_MODEL, api_key=settings.LLM_API_KEY, base_url=settings.LLM_BASE_URL)
    response=llm.invoke(prompt)
    return response.text()

print(introduce_company("Rikkeisoft", "English"))
````

# Structure Ouput in Langchain

````python
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

class Rikkeisoft(BaseModel):
    name: str = Field(description="The name of the company")
    year_founded: int = Field(description="The year the company was founded")
def get_company_info(company_name: str, language: str   ):
    template=PromptTemplate.from_template("Introduce about the company {company_name} in {language} language")
    prompt=template.format(company_name=company_name, language=language)
    llm=ChatOpenAI(model=settings.LLM_CHAT_MODEL, api_key=settings.LLM_API_KEY, base_url=settings.LLM_BASE_URL)
    response=llm.with_structured_output(Rikkeisoft)
    return response.invoke(prompt)

print(get_company_info("Rikkeisoft", "English"))
# Chat Prompt Template
````

# Ouput Parser And Langchain Expression Language(LCEL)

### Output Parser is a component that converts the output of a model into a structured format.


````python
from langchain_core.output_parsers import (
    JsonOutputParser, PydanticOutputParser, CommaSeparatedListOutputParser
)
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI
````

````python
# 1. JsonOutputParser - Parse to dict/JSON
parser = JsonOutputParser()
json_parser = JsonOutputParser()
structured = json_parser.parse('{"name": "LangChain", "stars": 80}')
print(structured)
print(type(structured))
````

````python
# 2. PydanticOutputParser - Validate with models
class Rikkeisoft(BaseModel):
    name: str = Field(description="The name of the company")
    year_founded: int = Field(description="The year the company was founded")   
llm=ChatOpenAI(model=settings.LLM_CHAT_MODEL, api_key=settings.LLM_API_KEY, base_url=settings.LLM_BASE_URL)
pydantic_parser = PydanticOutputParser(pydantic_object=Rikkeisoft)
prompt = ChatPromptTemplate.from_template(
    "Introduce about the company {company_name} in {language} language. Output in JSON format.\n{format_instructions}",
    partial_variables={"format_instructions": pydantic_parser.get_format_instructions()}
)
chain = prompt | llm | pydantic_parser
result = chain.invoke({"company_name": "Rikkeisoft", "language": "English"})
print(result)
````

````python
# 3. CommaSeparatedListOutputParser - Parse to list
csv_parser = CommaSeparatedListOutputParser()
result = csv_parser.parse("1. LangChain\n2. LangGraph\n3. LangSmith")
print(result)
````
